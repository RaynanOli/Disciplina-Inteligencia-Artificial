```python
#21.Busca Binária - Implemente a busca binária para encontrar um elemento em uma lista ordenada.

numeros = [1, 3, 5, 7, 9, 11, 13]
numero = int(input("Digite um número: "))

inicio = 0
fim = len(numeros) - 1

while inicio <= fim: # Continua a busca enquanto ainda houver elementos para verificar
    meio = (inicio + fim) // 2  # Encontra a posição do elemento do meio
    if numeros[meio] == numero:
        print("Número encontrado!")
        break
    elif numero > numeros[meio]:
        inicio = meio + 1 # Continua a busca na metade direita
    else:
        fim = meio - 1 # Continua a busca na metade esquerda
else:
    print("Número não encontrado.")
```

```python
#22. Criptografia de Cifra de César - Crie uma função para criptografar uma mensagem usando a cifra deCésar.

mensagem = input("Digite uma mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

mensagem_criptografada = ""
alfabeto = "abcdefghijklmnopqrstuvwxyz"

for letra in mensagem:
    if letra in alfabeto:
        posicao = alfabeto.index(letra) # Encontra a posição da letra no alfabeto
        nova_posicao = (posicao + deslocamento) % 26 # Calcula a nova posição da letra
        mensagem_criptografada += alfabeto[nova_posicao] # Adiciona a letra criptografada
    else:
        mensagem_criptografada += letra # Mantém espaços e outros caracteres
print("Mensagem criptografada:", mensagem_criptografada)
```

```python
#30.Jogo da Forca - Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra.

palavra = "bob"
tentativas = 6
letras_acertadas = []  # Guarda as letras que o jogador acertou

while tentativas > 0: # Mantém o jogo enquanto ainda houver tentativas
    palavra_mostrada = ""
    for letra in palavra: # Percorre cada letra da palavra
        if letra in letras_acertadas:
            palavra_mostrada += letra
        else:
            palavra_mostrada += "_"
    print("Palavra:", palavra_mostrada)

    if "_" not in palavra_mostrada: # Verifica se todas as letras foram descobertas
        print("Você ganhou!")
        break
    letra = input("Digite uma letra: ")

    if letra in palavra:
        letras_acertadas.append(letra) # Adiciona a letra quando o jogador acerta
        print("Você acertou!")
    else:
        tentativas -= 1 # Diminui uma tentativa quando o jogador erra
        print("Você errou! Tentativas restantes:", tentativas)

if tentativas == 0:
    print("Você perdeu! A palavra era:", palavra)
```

```python
#31.Ordenação por Bolha - Implemente o algoritmo de ordenação por bolha (Bubble Sort).

numeros = [5, 3, 9, 8, 0]

for i in range(len(numeros)): # Repete o processo para organizar toda a lista
    for x in range(len(numeros) - 1): # Compara os números que estão lado a lado
        if numeros[x] > numeros[x + 1]:
            numeros[x], numeros[x + 1] = numeros[x + 1], numeros[x] # Troca os números de posição
print("Lista ordenada:", numeros)
```

```python
#37. Funções Recursivas - Implemente uma função recursiva para calcular o fatorial de um número.

def fatorial(numero):
    if numero == 0: # Para a função quando o número chega a zero
        return 1
    else:
        return numero * fatorial(numero - 1) # Chama a função novamente com o número menor

numero = int(input("Digite um número: "))
resultado = fatorial(numero)
print("O fatorial é:", resultado)
```

```python
#48.Ordenação por Inserção - Implemente o algoritmo de ordenação por inserção (Insertion Sort).

numeros = [5, 2, 4, 1, 3]

for i in range(1, len(numeros)): # Percorre a lista a partir do segundo número
    numero = numeros[i] # Guarda o número que será colocado na posição correta
    j = i - 1
    while j >= 0 and numeros[j] > numero: # Compara o número com os anteriores
        numeros[j + 1] = numeros[j] # Move o número maior uma posição para frente
        j -= 1
    numeros[j + 1] = numero # Coloca o número na posição correta
print("Lista ordenada:", numeros)
```

```python
#50. Cálculo de Frequência de Letras - Calcule a frequência de cada letra em uma string fornecida pelo usuário.

texto = input("Digite uma palavra ou frase: ")
contagem = {} # Cria um dicionário vazio para guardar as letras e suas quantidades

for letra in texto: # Percorre cada letra do texto
    if letra.isalpha():
        if letra in contagem:
            contagem[letra] += 1
        else:
            contagem[letra] = 1
print("Frequência das letras:", contagem)
```
