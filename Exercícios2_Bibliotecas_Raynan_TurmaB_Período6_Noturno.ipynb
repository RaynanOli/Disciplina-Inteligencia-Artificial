````python
#19- Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

import numpy as np

array1 = np.random.randint(1, 11, 20) # Array com 20 números inteiros aleatórios
array2 = np.random.randint(1, 11, 20) # Array2 com 20 números inteiros aleatórios

produto_escalar = np.dot(array1, array2) # Calcula o produto escalar entre os dois arrays

norma1 = np.linalg.norm(array1) # Calcula o tamanho do primeiro array
norma2 = np.linalg.norm(array2) # Calcula o tamanho do segundo array

cos_angulo = produto_escalar / (norma1 * norma2) # Calcula o cosseno do ângulo entre os arrays

angulo = np.arccos(cos_angulo) # Calcula o ângulo em radianos a partir do cosseno
angulo_graus = np.degrees(angulo) # Converte o ângulo de radianos para graus
print("Array 1:", array1)
print("Array 2:", array2)
print("Ângulo:", angulo_graus, "graus")

# Calcula o produto escalar e o tamanho dos dois arrays #
# Depois usa esses valores para encontrar o cosseno e descobrir o ângulo entre eles em graus #
````

# **Exercícios de SciPy**

````python
#1- Calcule a integral da função f(x)= x²sin(x) no intervalo de 0 a π.

import numpy as np
from scipy.integrate import quad

def f(x):
    return x**2 * np.sin(x) # Define a função x²sen(x)
resultado, erro = quad(f, 0, np.pi) # Calcula a integral da função no intervalo de 0 até π
print("Resultado da integral:", resultado)
````

````python
#2- Resolva a equação diferencial dy/dx= x−y, com y(0)=1, no intervalo de 0 a 10.

from scipy.integrate import solve_ivp

def equacao(t, y):
    return t - y  # Define a equação diferencial dy/dx = x - y
solucao = solve_ivp(equacao, [0, 10], [1]) # Resolve a equação de 0 até 10, começando com y(0) = 1
print("Solução:", solucao.y[0])
````

````python
#3- Encontre as raízes da função f(x)= x⁴−3x³+2 utilizando SciPy.

from scipy.optimize import root

def funcao(x):
    return x**4 - 3*x**3 + 2 # Define a função x⁴ - 3x³ + 2
raiz = root(funcao, [0, 1, 2, 3]) # Encontrar as raízes usando valores iniciais
print("Raízes:", raiz.x)
````

````python
#4- Use a função de otimização de SciPy para maximizar f(x)= −x²+ 4x+ 1.

from scipy.optimize import minimize

def funcao(x):
    return -( -x**2 + 4*x + 1) # Inverte a função para encontrar o máximo
resultado = minimize(funcao, 0) # Procura o menor valor começando pelo valor 0
x_max = resultado.x[0]
valor_max = -resultado.fun
print("Máximo encontrado em x:", x_max)
print("Valor máximo:", valor_max)

## minimize() procura o menor valor ##
## Para encontrar o máximo de uma função usando minimize(), inverte o sinal da função e depois volta ao valor original ##
````

````python
#5- Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

import numpy as np
from scipy.fft import fft

t = np.linspace(0, 1, 500) # 500 pontos igualmente espaçados entre 0 e 1
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t) # Cria sinal com seno e cosseno
frequencias = fft(sinal) # Calcula a transformada de Fourier do sinal
print("Transformada de Fourier:", frequencias[:10]) # Print com 10 valores para ficar mais fácil o entendimento
print("\n")
print("Transformada de Fourier:", frequencias)
````

````python
#6- Encontre a matriz inversa de uma matriz 5x5 de números aleatórios.

import numpy as np
from scipy.linalg import inv

matriz = np.random.rand(5, 5) # Matriz 5x5 com números aleatórios entre 0 e 1
inversa = inv(matriz) # Calcula a matriz inversa
print("Matriz original:")
print(matriz)
print("\n")
print("\nMatriz inversa:")
print(inversa)
````

````python
#7- Use a função de interpolação cúbica de SciPy para interpolar os dados (0, 0),(1, 2),(2, 3),(3, 5),(4, 4).

import numpy as np
from scipy.interpolate import interp1d

x = np.array([0, 1, 2, 3, 4]) # Valores x
y = np.array([0, 2, 3, 5, 4]) # Valores y
f = interp1d(x, y, kind='cubic') # Função de interpolação cúbica
x_novos = np.array([0.5, 1.5, 2.5, 3.5]) # Pontos para estimar os valores
y_novos = f(x_novos) # Calcula os valores interpolados
print("Valores interpolados:", y_novos)
````

````python
#8-  Resolva o sistema de equações 3x + 2y = 5 e x + 4y = 6.

import numpy as np
from scipy.linalg import solve

A = np.array([[3, 2], [1, 4]]) # Coeficiente de x e y
b = np.array([5, 6]) # Valores do lado direito das equações
resultado = solve(A, b) # Resolve as equações
print("Valor de x:", resultado[0])
print("Valor de y:", resultado[1])
````

````python
#9- Ajuste uma curva quadr´atica aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8).

import numpy as np
from scipy.optimize import curve_fit

x = np.array([1, 2, 3, 4, 5]) # Valores de x
y = np.array([2, 4, 5, 7, 8]) # Valores de y
def funcao(x, a, b, c):
    return a * x**2 + b * x + c # Função quadrática
parametros, _ = curve_fit(funcao, x, y) # Encontra os valores de a, b e c
a, b, c = parametros
print("a =", a)
print("b =", b)
print("c =", c)
````

````python
#6- Ordene o DataFrame pela coluna Produto em ordem alfabética.

import pandas as pd

dados = {
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'],
    'Preço': [9900, 8, 1540, 120, 450],
    'Quantidade': [2, 10, 5, 3, 7]
}
df = pd.DataFrame(dados)
df_ordenado = df.sort_values(by='Produto') # Ordena coluna Produto em ordem alfabética
print(df_ordenado)
````

````python
#7- Crie uma nova coluna chamada Total que seja o produto entre as colunas Preço e Quantidade.

import pandas as pd

dados = {  # Cria o DataFrame
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'],
    'Preço': [3500, 80, 150, 1200, 250],
    'Quantidade': [2, 10, 5, 3, 7]
}
df = pd.DataFrame(dados)
df['Total'] = df['Preço'] * df['Quantidade']

print(df)
````

````python
#8- Agrupe o DataFrame pela coluna Categoria e calcule o total de produtos por categoria.

import pandas as pd

dados = {
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'],
    'Categoria': ['Eletrônicos', 'Acessórios', 'Acessórios', 'Eletrônicos', 'Acessórios'],
    'Preço': [3500, 80, 150, 1200, 250],
    'Quantidade': [2, 10, 5, 3, 7]
}

df = pd.DataFrame(dados)
total_categoria = df.groupby('Categoria')['Quantidade'].sum() # Agrupa por categoria e calcular o total
print(total_categoria)
````

````python
#11- Preencha os valores nulos de uma coluna numérica com a mediana dessa coluna.

import pandas as pd

dados = {
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'],
    'Preço': [None, 10, None, None, 1000]
}
df = pd.DataFrame(dados)
mediana = df['Preço'].median() # Calcula mediana
df['Preço'] = df['Preço'].fillna(mediana) # Pega os valores nulos e coloca a mediana
print(df)
````

````python
#12-  Verifique quais colunas de um DataFrame contêm valores duplicados

import pandas as pd

dados = {
    'Produto': ['Notebook', 'Mouse', 'Monitor', 'Monitor', 'Headset'],
    'Preço': [3500, 80, 80, 1200, 250],
    'Quantidade': [2, 10, 10, 3, 7]
}
df = pd.DataFrame(dados)
for coluna in df.columns: # Verifica quais colunas possuem valores duplicados
    if df[coluna].duplicated().any():
        print(f"A coluna '{coluna}' contém valores duplicados.")
````

````python
#13- Crie uma tabela dinâmica a partir do DataFrame, calculando a soma de Quantidade por Categoria.

import pandas as pd

dados = {
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'],
    'Categoria': ['Eletrônicos', 'Acessórios', 'Acessórios', 'Eletrônicos', 'Acessórios'],
    'Preço': [3500, 80, 150, 1200, 250],
    'Quantidade': [2, 10, 5, 3, 7]
}
df = pd.DataFrame(dados)
tabela = df.pivot_table( # Cria tabela dinamica
    values='Quantidade',
    index='Categoria',
    aggfunc='sum'
)
print(tabela)
````

````python
#14- Mescle dois DataFrames baseados em uma coluna comum chamada ID utilizando um join.

import pandas as pd

dados1 = { # Dataframe 1
    'ID': [1, 2, 3, 4],
    'Nome': ['Bob', 'Yuri22', 'Dona', 'Fabio']
}
df1 = pd.DataFrame(dados1)
dados2 = { # Dataframe 2
    'ID': [1, 2, 3, 4],
    'Salário': [5000, 6000, 7000, 8000]
}
df2 = pd.DataFrame(dados2)
df_final = pd.merge(df1, df2, on='ID') # Junta os 2 dataframes usando coluna IDD
print(df_final)
````

````python
#15- Crie um DataFrame com dados faltantes e use a interpolação linear para preenchê-los.

import pandas as pd

dados = {
    'Produto': ['A', 'B', 'C', 'D', 'E'],
    'Preço': [100, None, 400, None, 800]
}
df = pd.DataFrame(dados)
df['Preço'] = df['Preço'].interpolate(method='linear') # Preenche os NONE com interpolação linear
print(df)
````

````python
#19- Crie um DataFrame a partir de um dicionário de listas e calcule a média de cada coluna numérica.

import pandas as pd

dados = {
    'Idade': [20, 25, 30, 35, 40],
    'Salário': [2000, 3000, 4000, 5000, 6000],
    'Horas': [6, 7, 8, 9, 10]
}
df = pd.DataFrame(dados)
medias = df.mean() # Calcula media de cada coluna numerica
print(medias)
````

````python
#20- Agrupe os dados por uma coluna categ´orica e calcule a mediana de uma coluna numérica.

import pandas as pd

dados = {
    'Categoria': ['Eletrônico', 'Eletrônico', 'Roupa', 'Roupa', 'Eletrônico', 'Roupa'],
    'Preço': [100, 200, 1, 80, 300, 100]
}
df = pd.DataFrame(dados)
mediana = df.groupby('Categoria')['Preço'].median() # Agrupa por categoria e calcular a mediana do preço
print(mediana)
````

## 🔹 Importação e criação de DataFrames

| Função / Recurso      | O que faz                   |
| --------------------- | --------------------------- |
| `import pandas as pd` | Importa a biblioteca Pandas |
| `pd.DataFrame()`      | Cria um DataFrame           |
| `pd.Series()`         | Cria uma Series             |
| `df.columns`          | Mostra os nomes das colunas |
| `df.head()`           | Exibe as primeiras linhas   |
| `df.tail()`           | Exibe as últimas linhas     |

---

## 🔹 Leitura e gravação de arquivos

| Função / Recurso  | O que faz                     |
| ----------------- | ----------------------------- |
| `pd.read_csv()`   | Lê um arquivo CSV             |
| `pd.read_excel()` | Lê um arquivo Excel           |
| `pd.read_json()`  | Lê um arquivo JSON            |
| `pd.read_sql()`   | Lê dados de uma consulta SQL  |
| `df.to_csv()`     | Exporta um DataFrame para CSV |

---

## 🔹 Seleção de dados

| Função / Recurso             | O que faz                                 |
| ---------------------------- | ----------------------------------------- |
| `df['Coluna']`               | Seleciona uma coluna                      |
| `df[['Coluna1', 'Coluna2']]` | Seleciona várias colunas                  |
| `df.iloc[]`                  | Seleciona dados usando posições numéricas |
| `df.loc[]`                   | Seleciona dados usando índices/rótulos    |
| `df[condição]`               | Filtra linhas de acordo com uma condição  |

---

## 🔹 Adição e remoção de colunas

| Função / Recurso       | O que faz                  |
| ---------------------- | -------------------------- |
| `df['Nova'] = valores` | Cria uma nova coluna       |
| `df.drop()`            | Remove colunas ou linhas   |
| `axis=1`               | Indica operação em colunas |
| `axis=0`               | Indica operação em linhas  |

---

## 🔹 Filtros e condições

| Recurso                  | O que faz                      |                                |
| ------------------------ | ------------------------------ | ------------------------------ |
| `df[df['Preço'] > 100]`  | Filtra valores maiores que 100 |                                |
| `df[df['Preço'] < 100]`  | Filtra valores menores que 100 |                                |
| `df[df['Preço'] == 100]` | Filtra valores iguais a 100    |                                |
| `&`                      | Representa **E** em condições  |                                |
| `                        | `                              | Representa **OU** em condições |
| `>`                      | Maior que                      |                                |
| `<`                      | Menor que                      |                                |
| `==`                     | Igual a                        |                                |

## 🔹 Estatística

| Função / Recurso | O que faz              |
| ---------------- | ---------------------- |
| `.mean()`        | Calcula a média        |
| `.median()`      | Calcula a mediana      |
| `.sum()`         | Calcula a soma         |

| `.min()`         | Encontra o menor valor |

## 🔹 Agrupamento de dados

| Função / Recurso | O que faz                         |
| ---------------- | --------------------------------- |
| `groupby()`      | Agrupa os dados por uma categoria |
| `.mean()`        | Calcula a média de cada grupo     |
| `.median()`      | Calcula a mediana de cada grupo   |
| `.sum()`         | Soma os valores de cada grupo     |

### 💡 Exemplo

```python
df.groupby('Categoria')['Preço'].median()
```

→ Agrupa os produtos por **Categoria** e calcula a **mediana do Preço** de cada categoria.

### 🧠 Lógica para decorar

**GROUPBY → AGRUPAR → ESCOLHER COLUNA → CALCULAR**

```python
df.groupby('Categoria')['Preço'].mean()
```

## 🔹 Dados ausentes

| Função / Recurso | O que faz                                         |
| ---------------- | ------------------------------------------------- |
| `isnull()`       | Identifica valores nulos                          |
| `dropna()`       | Remove linhas/colunas com valores nulos           |
| `fillna()`       | Preenche valores nulos                            |
| `.mean()`        | Pode ser usada para preencher nulos com a média   |
| `.median()`      | Pode ser usada para preencher nulos com a mediana |
| `interpolate()`  | Preenche valores faltantes por interpolação       |

### 🗑️ Remover valores nulos

```python
df = df.dropna()
```

→ Remove as linhas que possuem valores nulos.

### 🔄 Preencher pela mediana

```python
df['Coluna'] = df['Coluna'].fillna(df['Coluna'].median())
```

→ Substitui valores nulos pela mediana da coluna.

### 📈 Interpolação linear

```python
df['Coluna'] = df['Coluna'].interpolate()
```

→ Estima os valores faltantes com base nos valores existentes ao redor.

## 🔹 Pivot Table

| Função / Recurso | O que faz                              |
| ---------------- | -------------------------------------- |
| `pivot_table()`  | Cria uma tabela dinâmica               |

| `aggfunc=`       | Define a operação utilizada            |

### 💡 Exemplo

```python
df.pivot_table(
    values='Quantidade',
    index='Categoria',
    aggfunc='sum'

```

→ Calcula a **soma da Quantidade por Categoria**.

```text
mean → média

sum → soma
count → quantidade
```

````python
#1- Crie um gráfico de linha para a função y = x³ no intervalo de x = −10 até x = 10.

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10, 10, 100) # Gera valor de x
y = x**3 # Calcula y
plt.plot(x, y) # Cria gráfico de linha
plt.title("Gráfico de Linha - Função x³")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
````

````python
#2- Plote um gráfico de dispersão para 100 pontos aleatórios com uma paleta de cores baseada nos valores de y.

import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(100) # 100 pontos aleatorios x
y = np.random.rand(100) # 100 pontos aleatorios y
plt.scatter(x, y, c=y, cmap="viridis") # Cria gráfico de dispersão
plt.title("Gráfico de Dispersão - 100 Pontos")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
````

→ Gráfico de proporções.


→ Distribuição dos dados.


→ Cria uma posição para um gráfico dentro da figura.

